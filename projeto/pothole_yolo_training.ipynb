## 🚀 Treinamento do modelo YOLOv8

````python
MODEL_PATH = 'Models/yolov8n.pt'
DATA_PATH = 'PD/data.yaml' 
````

Verificação rápida antes do treino

````python
print("yolov8n.pt:", os.path.exists(MODEL_PATH))

print("data.yaml:", os.path.exists(DATA_PATH))
````

````python
model = YOLO(MODEL_PATH)
````

````python
print("CUDA disponível:", torch.cuda.is_available())
print("Número de GPUs:", torch.cuda.device_count())

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
````

Isso te dá o máximo teórico de threads

````python
print("CPUs lógicas:", os.cpu_count())
````

````python
model.train(
    data=DATA_PATH,
    epochs=100,
    imgsz=416,
    batch=32,
    device=0,
    workers=0,
    cache='ram',
    amp=True,
    cos_lr=True,
    patience=30
)
````

## 📊 Validate model

````python
# 📊 Validate trained model
model.val()
````

## 🖼️ Test inference on image

````python
# 🖼️ Run prediction on sample image
results = model('test2.jpg', show=True)
print(results)
````

## ⚡ Export model for Android (TFLite)

````python
# 📦 Export model to TFLite (Android)
model.export(
    format='tflite',
    int8=True  # ⚡ melhor performance para mobile
)
````
