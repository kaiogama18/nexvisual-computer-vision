```python
df = pd.read_csv("data/moma/artworks.csv")
```

```python
artist_counts = df["Artist ID"].value_counts()
```

```python
selected_artists = artist_counts[
    artist_counts >= 20
].index

df_model = df[
    df["Artist ID"].isin(selected_artists)
].copy()

print(df_model.shape)
print(df_model["Name"].nunique())
```
